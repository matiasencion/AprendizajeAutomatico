```python
import sys
from pathlib import Path

# El notebook se ejecuta desde Tarea1/notebooks; el código está en ../src.
sys.path.insert(0, str(Path("../src").resolve()))

import pandas as pd
from load import load_base_dataset

dataset = load_base_dataset("../datos/futbol_uruguayo.csv")

print(dataset.head())
```

```python
import sys
import os

# Agrega la carpeta padre (Tarea1) al path de búsqueda de Python

from baseClassifier.base_classifier import base_classifier

row = dataset.iloc[3000]
result = base_classifier(dataset, 10, row)
print(result)
```

```python
test = dataset[
    (dataset["date"] >= pd.Timestamp("2024-01-01"))
    & (dataset["date"] < pd.Timestamp("2026-01-01"))
].copy()

print("Cantidad de partidos de evaluacion:", len(test))
```

```python
y_true = []
y_pred = []

for _, row in test.iterrows():

    # Real result
    actual = row["result"]

    # Prediction
    prediction = base_classifier(dataset, 10, row)

    y_true.append(actual)
    y_pred.append(prediction)
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

accuracy = accuracy_score(y_true, y_pred)

balanced_accuracy = balanced_accuracy_score(y_true, y_pred)

f1 = f1_score(
    y_true,
    y_pred,
    average="macro"
)

print("Accuracy:", accuracy)
print("Balanced accuracy:", balanced_accuracy)
print("Macro F1:", f1)

print("\nClassification report:")
print(classification_report(y_true, y_pred))

print("\nConfusion matrix:")
print(confusion_matrix(y_true, y_pred))
```
